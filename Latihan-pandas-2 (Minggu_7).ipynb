```python
#import data csv dari lokal
import pandas as pd
df_data = pd.read_csv(r'D:\Michael Road To Data Analyst\Dataset\SampleSuperstore.csv')
df_data.head()
```

```python
#Membuat data frame baru dan belajar untuk menggabungkan data frame dengan merge
df_baru = pd.DataFrame({'Region': ['Central', 'East', 'South', 'West'],
                        'Region_Manager': ['John Smith', 'Jane Doe', 'Michael Johnson', 'Emily Davis']})
gabungan_df = pd.merge(df_data, df_baru, on='Region', how='left')
print(gabungan_df)
```

```python
#Belajar menggabungkan data frame secara bertumpuk dengan concat
tabel_1 = gabungan_df.iloc[:4996]
print(tabel_1)
tabel_2 = gabungan_df.iloc[4996:]
print(tabel_2)
gabung_lagi = pd.concat([tabel_1, tabel_2])
print(gabung_lagi)
```

```python
#Belajar membuat pivot table
sales_by_region = pd.pivot_table(gabung_lagi, index = 'Region', values = 'Sales', aggfunc = 'sum')
print(sales_by_region)
profit_by_category = pd.pivot_table(gabung_lagi, index = 'Category', values = 'Profit', aggfunc = 'sum')
print(profit_by_category)
volume_of_transaction_by_segment = pd.pivot_table(gabung_lagi, index = 'Segment', values = 'Sales', aggfunc = 'count')
print(volume_of_transaction_by_segment)
```

```python
#Belajar mendeteksi missing value
df_data.Quantity = df_data.Quantity.replace(2,None)
df_data.isnull().sum()
```

```python
#Belajar mengisi missing value dengan metode fillna
df_data.Quantity = df_data.Quantity.fillna(2)
df_data.isnull().sum()
```

```python
#Belajar mengelompokkan data dengan groupby
df_data.groupby(['Ship Mode', 'Segment'])['Profit'].agg(['sum', 'mean', 'count'])
```
